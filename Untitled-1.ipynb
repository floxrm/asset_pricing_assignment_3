## Numerical exercise: evaluating four mutual funds

| Ticker | Fund |
|---|---|
| TRBCX | T. Rowe Price Blue Chip Growth |
| GLCGX | Goldman Sachs Large Cap Growth Insights |
| STSEX | BlackRock Exchange Portfolio (BlackRock Shares) |
| DFSVX | DFA U.S. Small Cap Value Portfolio |

**Sample:** monthly returns, January 2002 – July 2026 (last price date 2026-07-01).
**Data:** Yahoo Finance (adjusted fund prices) and the Kenneth French data library (Fama–French 3 factors, momentum, risk-free rate).

```python
# If yfinance is missing, run once:  %pip install yfinance
import io
import re
import zipfile

import numpy as np
import pandas as pd
import requests
import statsmodels.api as sm
import yfinance as yf

FUNDS = {
    "TRBCX": "T. Rowe Price Blue Chip Growth",
    "GLCGX": "Goldman Sachs Large Cap Growth Insights",
    "STSEX": "BlackRock Exchange Portfolio",
    "DFSVX": "DFA U.S. Small Cap Value",
}
TICKERS = list(FUNDS)

PRICE_START = "2001-12-01"  # one month early, so January 2002 has a return
PRICE_END = "2026-07-02"    # yfinance's end date is exclusive -> last bar is 2026-07-01
SAMPLE_START, SAMPLE_END = "2002-01", "2026-07"

FRENCH_URL = "https://mba.tuck.dartmouth.edu/pages/faculty/ken.french/ftp/"
ANN = 12  # months per year, used to annualize
```

---
## 1. Data and monthly excess returns

### 1a. Fund prices from Yahoo Finance
We use **adjusted** closing prices (`auto_adjust=True`). Mutual funds pay out dividends and capital-gain distributions, which make the raw price drop. Adjusted prices add these payouts back, so `pct_change()` gives the **total return**.

Yahoo's monthly bar dated the 1st of month *t* holds the **closing price at the end of month *t***. The December 2001 bar is therefore the starting value for the January 2002 return.

```python
prices = yf.download(
    TICKERS,
    start=PRICE_START,
    end=PRICE_END,
    interval="1mo",
    auto_adjust=True,
    progress=False,
)["Close"][TICKERS]
prices.index = prices.index.to_period("M")

fund_ret = prices.pct_change().dropna()

print(f"Prices : {prices.index[0]} -> {prices.index[-1]}")
print(f"Returns: {fund_ret.index[0]} -> {fund_ret.index[-1]}  ({len(fund_ret)} months)")
fund_ret.head()
```

### 1b. Risk-free rate and factors from the Kenneth French data library
We download the zipped CSV files directly (`pandas_datareader` doesn't work with pandas 3):

* `F-F_Research_Data_Factors` → `Mkt-RF`, `SMB`, `HML`, `RF`
* `F-F_Momentum_Factor` → `Mom` (the Carhart momentum factor; "FFC" = Fama–French–Carhart)

Each file has a **monthly** block (dates as `YYYYMM`) followed by an **annual** block. We keep only the monthly rows. The values are in **percent**, so we divide by 100.

```python
MONTHLY_ROW = re.compile(r"^\s*\d{6}\s*,")  # a line that starts with YYYYMM


def load_french_monthly(zip_name):
    # Download one Kenneth French CSV zip and return its monthly table in decimals.
    response = requests.get(FRENCH_URL + zip_name, timeout=60)
    response.raise_for_status()
    with zipfile.ZipFile(io.BytesIO(response.content)) as z:
        lines = z.read(z.namelist()[0]).decode("latin-1").splitlines()

    first = next(i for i, line in enumerate(lines) if MONTHLY_ROW.match(line))
    columns = [c.strip() for c in lines[first - 1].split(",")[1:]]  # header sits right above

    rows = []
    for line in lines[first:]:
        if not MONTHLY_ROW.match(line):
            break  # the monthly block ends where the annual block starts
        date, *values = [v.strip() for v in line.split(",")]
        rows.append([date, *map(float, values)])

    table = pd.DataFrame(rows, columns=["date", *columns])
    table.index = pd.PeriodIndex(table.pop("date"), freq="M")
    return table / 100


ff3 = load_french_monthly("F-F_Research_Data_Factors_CSV.zip")  # Mkt-RF, SMB, HML, RF
mom = load_french_monthly("F-F_Momentum_Factor_CSV.zip")        # Mom
factors = ff3.join(mom, how="inner")
factors.loc[SAMPLE_START:SAMPLE_END].tail()
```

### 1c. Merge and compute excess returns
$$r^e_{i,t} = r_{i,t} - r_{f,t}$$

The market's excess return is French's `Mkt-RF` (value-weighted CRSP market minus the T-bill rate).

```python
data = fund_ret.join(factors, how="inner").loc[SAMPLE_START:SAMPLE_END]

expected_months = len(pd.period_range(SAMPLE_START, SAMPLE_END, freq="M"))
assert len(data) == expected_months, "missing months after merging"
assert not data.isna().any().any(), "missing values in the merged data"

excess = data[TICKERS].sub(data["RF"], axis=0)  # fund excess returns
mkt_excess = data["Mkt-RF"]                     # market excess return

print(f"Sample: {data.index[0]} -> {data.index[-1]}  ({len(data)} months)")
excess.head()
```

```python
# Quick look: annualized mean and volatility of the monthly excess returns (in %)
pd.DataFrame(
    {
        "Mean excess return (%/yr)": excess.mean() * ANN * 100,
        "Volatility (%/yr)": excess.std() * np.sqrt(ANN) * 100,
        "Min month (%)": excess.min() * 100,
        "Max month (%)": excess.max() * 100,
    }
).round(2)
```

---
## 2. Sharpe ratio and Information ratio

**Sharpe ratio:** excess return per unit of total risk
$$SR_i = \frac{E[r^e_i]}{\sigma(r^e_i)}$$

**Information ratio relative to the market:** the active return over the market per unit of active risk (the *tracking error*)
$$IR_i = \frac{E[r_i - r_M]}{\sigma(r_i - r_M)}, \qquad r_i - r_M = r^e_i - r^e_M$$

Everything is annualized: means × 12, standard deviations × √12, so both ratios get × √12.

As a robustness check we also report the **appraisal ratio**. Some textbooks (e.g. BKM) call this the information ratio:
$$AR_i = \frac{\alpha_i}{\sigma(\varepsilon_i)} \quad\text{from the CAPM regression}\quad r^e_{i,t} = \alpha_i + \beta_i\, r^e_{M,t} + \varepsilon_{i,t}$$
The tracking-error IR implicitly assumes β = 1. The appraisal ratio adjusts for the fund's actual market beta.

```python
def annualized_ratio(series):
    return series.mean() / series.std() * np.sqrt(ANN)


active = excess.sub(mkt_excess, axis=0)  # r_i - r_M

capm_alpha, capm_beta, capm_resid_vol = {}, {}, {}
for fund in TICKERS:
    capm = sm.OLS(excess[fund], sm.add_constant(mkt_excess)).fit()
    capm_alpha[fund] = capm.params["const"]
    capm_beta[fund] = capm.params["Mkt-RF"]
    capm_resid_vol[fund] = np.sqrt(capm.mse_resid)
capm_alpha, capm_beta, capm_resid_vol = map(pd.Series, (capm_alpha, capm_beta, capm_resid_vol))

performance = pd.DataFrame(
    {
        "Mean excess (%/yr)": excess.mean() * ANN * 100,
        "Volatility (%/yr)": excess.std() * np.sqrt(ANN) * 100,
        "Sharpe ratio": excess.apply(annualized_ratio),
        "Active return (%/yr)": active.mean() * ANN * 100,
        "Tracking error (%/yr)": active.std() * np.sqrt(ANN) * 100,
        "Information ratio": active.apply(annualized_ratio),
        "CAPM beta": capm_beta,
        "CAPM alpha (%/yr)": capm_alpha * ANN * 100,
        "Appraisal ratio": capm_alpha / capm_resid_vol * np.sqrt(ANN),
    }
)
performance.loc["Market"] = {
    "Mean excess (%/yr)": mkt_excess.mean() * ANN * 100,
    "Volatility (%/yr)": mkt_excess.std() * np.sqrt(ANN) * 100,
    "Sharpe ratio": annualized_ratio(mkt_excess),
    "CAPM beta": 1.0,
}
performance.round(3)
```

### Interpretation (Part 2)

**Sharpe ratios.** No fund beats the market on a total-risk basis. The market's Sharpe ratio is 0.61. The funds range from 0.50 (DFSVX) to 0.58 (TRBCX, STSEX). DFSVX has the highest average excess return (10.7%/yr) but also by far the highest volatility (21.5%/yr), so it has the *lowest* Sharpe ratio.

**Information ratios (vs. the market).** All four are small in absolute value (|IR| < 0.3; an IR around 0.5 is usually considered very good):

* **DFSVX (+0.12)** and **TRBCX (+0.09)** beat the market on average (+1.4% and +0.5% per year). Both carry large tracking errors, especially DFSVX at 11.3%/yr, because a small-value fund looks very different from the market index.
* **GLCGX (−0.17)** and **STSEX (−0.29)** trail the market.

**Why the appraisal ratio tells a different story.** The tracking-error IR treats every deviation from the market as "active", including simply having a market beta different from 1. Using the CAPM, the active return splits exactly into
$\bar r_i - \bar r_M = \alpha_i + (\beta_i - 1)\,\bar r^e_M$:

* DFSVX has β ≈ 1.2. Its higher return is compensation for extra market risk, and its CAPM alpha is actually *negative* (−0.6%/yr).
* STSEX has β ≈ 0.8. It trails the market mainly because it takes less market risk. Its CAPM alpha is slightly *positive* (+0.1%/yr).
* All appraisal ratios are close to zero (−0.23 to +0.03). After adjusting for market beta, no fund earns a meaningful abnormal return. Part 3 checks whether this still holds once we also control for size, value and momentum exposure.

## 3. Is there evidence of manager skill? Factor regressions

We regress each fund's monthly excess return on the factors:

**Fama–French 3-factor (FF3):**
$$r^e_{i,t} = \alpha_i + \beta_i\,(r_M - r_f)_t + s_i\, SMB_t + h_i\, HML_t + \varepsilon_{i,t}$$

**Fama–French–Carhart 4-factor (FFC4):**
$$r^e_{i,t} = \alpha_i + \beta_i\,(r_M - r_f)_t + s_i\, SMB_t + h_i\, HML_t + m_i\, MOM_t + \varepsilon_{i,t}$$

The factor loadings capture the return the fund earns from its *style* (market, size, value, momentum), which an investor could get cheaply from passive factor portfolios. **Skill shows up in the intercept α.** It is the average return the factors cannot explain.

* **Test:** $H_0: \alpha_i = 0$ vs. $H_1: \alpha_i \neq 0$. Evidence of skill requires α > 0 **and** statistical significance (|t| > 1.96 at the 5% level).
* **Standard errors:** Newey–West (HAC), robust to heteroskedasticity and autocorrelation. The lag length follows the rule of thumb $\lfloor 4(T/100)^{2/9} \rfloor$. Plain OLS t-statistics are shown for comparison.
* Significance stars: \* 10%, \*\* 5%, \*\*\* 1%.

```python
MODELS = {
    "FF3": ["Mkt-RF", "SMB", "HML"],
    "FFC4": ["Mkt-RF", "SMB", "HML", "Mom"],
}
T = len(data)
NW_LAGS = int(np.floor(4 * (T / 100) ** (2 / 9)))
print(f"T = {T} months, Newey-West lags = {NW_LAGS}")


def fit_factor_model(fund, factor_cols, hac=True):
    X = sm.add_constant(data[factor_cols])
    model = sm.OLS(excess[fund], X)
    if hac:
        return model.fit(cov_type="HAC", cov_kwds={"maxlags": NW_LAGS})
    return model.fit()


fits = {
    (model, fund): fit_factor_model(fund, cols)
    for model, cols in MODELS.items()
    for fund in TICKERS
}
ols_fits = {
    (model, fund): fit_factor_model(fund, cols, hac=False)
    for model, cols in MODELS.items()
    for fund in TICKERS
}
```

```python
LABELS = {"const": "alpha (%/month)", "Mkt-RF": "beta MKT", "SMB": "s  SMB", "HML": "h  HML", "Mom": "m  MOM"}


def stars(p):
    return "***" if p < 0.01 else "**" if p < 0.05 else "*" if p < 0.10 else ""


def regression_table(model):
    # Coefficients with significance stars, Newey-West t-statistics in parentheses below.
    terms = ["const", *MODELS[model]]
    index = [label for term in terms for label in (LABELS[term], "")] + ["adj. R²", "N"]
    columns = {}
    for fund in TICKERS:
        res = fits[(model, fund)]
        cells = []
        for term in terms:
            scale = 100 if term == "const" else 1  # alpha in % per month
            cells += [f"{res.params[term] * scale:.3f}{stars(res.pvalues[term])}", f"({res.tvalues[term]:.2f})"]
        cells += [f"{res.rsquared_adj:.3f}", f"{int(res.nobs)}"]
        columns[fund] = cells
    return pd.DataFrame(columns, index=index)


print("Fama-French 3-factor model  (t-statistics, Newey-West, in parentheses)")
regression_table("FF3")
```

```python
print("Fama-French-Carhart 4-factor model  (t-statistics, Newey-West, in parentheses)")
regression_table("FFC4")
```

### Alpha summary: FF3 vs. FFC4

```python
alpha_summary = pd.concat(
    {
        model: pd.DataFrame(
            {
                "alpha (%/yr)": [fits[(model, f)].params["const"] * ANN * 100 for f in TICKERS],
                "t (NW)": [fits[(model, f)].tvalues["const"] for f in TICKERS],
                "t (OLS)": [ols_fits[(model, f)].tvalues["const"] for f in TICKERS],
                "p-value (NW)": [fits[(model, f)].pvalues["const"] for f in TICKERS],
            },
            index=TICKERS,
        )
        for model in MODELS
    },
    axis=1,
)
alpha_summary.round(3)
```

### Interpretation (Part 3)

**FF3 model: no evidence of skill.** None of the four alphas is statistically different from zero:

| Fund | α (%/yr) | t (NW) | Verdict |
|---|---|---|---|
| TRBCX | +0.04 | 0.05 | essentially zero |
| GLCGX | −1.08 | −1.61 | negative, not significant (p = 0.11) |
| STSEX | −0.13 | −0.18 | essentially zero |
| DFSVX | +0.07 | 0.10 | essentially zero |

**FFC4 model: same conclusion.** Adding momentum barely changes anything. The alphas move by at most 0.25 %-points per year, and the adjusted R² values are essentially unchanged. Still no fund has a significantly **positive** alpha. The only change is that **GLCGX's** alpha becomes slightly more negative (−1.28%/yr, t = −1.87, p = 0.06), which is significant at the 10% level only. GLCGX has a small positive momentum loading (m = 0.035, significant at 10%). Part of its return is therefore explained by momentum exposure, which leaves even less for the manager. This is weak evidence of **underperformance**, not skill.

**The factor loadings match the fund styles:**

* **TRBCX and GLCGX** (large-cap growth): β slightly above 1, negative SMB (large firms), strongly negative HML (growth stocks).
* **STSEX** (large-cap blend): β ≈ 0.85, strongly negative SMB (mega-caps), small positive HML.
* **DFSVX** (small-cap value): large positive SMB (0.85) and HML (0.64), exactly as the name says.
* Momentum loadings are small and not significant at 5% for any fund. That's why FF3 and FFC4 give almost the same alphas.

**The factors explain almost everything.** The adjusted R² is 0.89–0.97. These funds' returns are mostly their factor exposures, which an investor could replicate cheaply with passive factor/index portfolios. DFSVX (R² = 0.974) is essentially a systematic size-and-value tilt. Its positive information ratio in Part 2 came from SMB/HML exposure, not stock picking. Once we control for it, its alpha is about zero.

**Conclusion.** Neither the FF3 nor the FFC4 model shows evidence of manager skill (a significantly positive α) for any of the four funds over 2002–2026. Two caveats:

* Fund returns on Yahoo are **net of fees**. An alpha of about zero after fees means that, at best, the managers generate just enough gross alpha to cover their fees, so investors capture none of it (consistent with Berk & Green, 2004).
* We test four funds at once. With a multiple-testing correction, the stricter significance thresholds (e.g. |t| > 3 as proposed by Harvey, Liu & Zhu, 2016) make the evidence for any skill even weaker.
